```python
import pandas as pd
data=pd.read_csv("student_performance.csv")
data
```

```python
data.isnull().sum()
```

```python
data['Total_Score']=data['Math_Score']+data['Science_Score']+data['English_Score']
```

```python
data
```

```python
data['Percentage']=(data['Total_Score']/3)
data
```

```python
def grade_(Percentage):
    if Percentage > 90 :
        return 'A'
    elif Percentage <=90 and Percentage >80:
        return 'B'
    elif Percentage <=80 and Percentage >70:
        return 'C'
    elif Percentage <=70 and Percentage >36:
        return 'Pass'
    else:
        return 'F'
data['Grade']=data['Percentage'].apply(grade_)
```

```python
data
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

x=data.drop(['Grade','Name','Student_ID','Total_Score','Percentage'],axis=1)
y=data['Grade']

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)

model=LogisticRegression(max_iter=1000)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score
print(model.score(x_train,y_train))
print(model.score(x_test,y_test))

print(accuracy_score(y_test,y_pred))
```
